# Sistema de Gestión de Expedientes Médicos (Pila)

Algoritmo con una **pila (stack, LIFO)** para gestionar expedientes médicos.
Cada expediente nuevo se coloca **encima** de la pila (push). Para atender o revisar, siempre se toma el que está **arriba de todo** (pop).

Cada expediente es un diccionario con: `numero_expediente`, `nombre`, `edad`, `motivo_consulta`.

**Este notebook está dividido en celdas pequeñas**, cada una con su explicación, para que puedas entender qué hace cada parte del código antes de ejecutarlo todo junto al final.

## Operaciones del menú
1. **Añadir** expediente (push)
2. **Eliminar toda la pila** (vaciarla por completo)
3. **Consultar** (ver todos los expedientes en espera)
4. **Contar** (cuántos expedientes hay en la pila)
5. **Eliminar el elemento de la cima** (pop)
6. **Salir**


## Parte 1: La estructura de datos (la pila)

Usamos una **lista de Python** (`pila_expedientes`) para representar la pila. No necesitamos una clase especial: una lista ya se comporta como una pila si solo la modificamos por un extremo (el final).

- `pila_expedientes[-1]` siempre será la **cima** (el expediente más reciente).
- `contador_expedientes` es una variable aparte que solo sirve para darle un número único a cada expediente nuevo (1, 2, 3, ...), aunque después se eliminen expedientes de en medio.

```python
pila_expedientes = []
contador_expedientes = 0
```

## Parte 2: Añadir un expediente (push)

Esta función:
1. Pide al usuario los datos del paciente (`nombre`, `edad`, `motivo`).
2. Aumenta el contador para generar un `numero_expediente` único.
3. Arma un **diccionario** con esos 4 datos.
4. Usa `pila_expedientes.append(...)`, que en Python agrega el elemento **al final de la lista** — esto equivale a colocar el expediente **arriba de la pila**.

`global contador_expedientes` es necesario porque dentro de la función vamos a *modificar* una variable que fue creada fuera de ella (en el nivel principal del programa).

```python
def anadir_expediente():
    """Agrega un nuevo expediente a la cima de la pila (push)."""
    global contador_expedientes

    nombre = input("Nombre del paciente: ")
    edad = input("Edad: ")
    motivo = input("Motivo de consulta: ")

    contador_expedientes += 1
    expediente = {
        "numero_expediente": contador_expedientes,
        "nombre": nombre,
        "edad": edad,
        "motivo_consulta": motivo
    }

    pila_expedientes.append(expediente)
    numero = expediente["numero_expediente"]
    print("\nExpediente #" + str(numero) + " agregado correctamente.\n")
```

## Parte 3: Eliminar toda la pila

Esta función vacía la pila **por completo**, sin importar cuántos expedientes tenga.

- Primero revisa si ya está vacía, para no pedir confirmación innecesariamente.
- Pide una **confirmación** (`s`/`n`) antes de borrar, porque es una acción irreversible: una vez vaciada, esos datos ya no existen en el programa.
- `pila_expedientes.clear()` es el método de Python que elimina todos los elementos de una lista, dejándola vacía (`[]`), sin necesidad de crear una lista nueva.

```python
def eliminar_toda_la_pila():
    """Elimina todos los expedientes de la pila (vaciarla por completo)."""
    if len(pila_expedientes) == 0:
        print("\nLa pila ya esta vacia, no hay nada que eliminar.\n")
        return

    confirmacion = input("Seguro que deseas eliminar TODOS los expedientes? (s/n): ")
    if confirmacion.lower() == "s":
        pila_expedientes.clear()
        print("\nSe eliminaron todos los expedientes de la pila.\n")
    else:
        print("\nOperacion cancelada.\n")
```

## Parte 4: Consultar la pila

Esta función solo **muestra** información, no modifica la pila (por eso no la "atiende" ni la elimina).

- `reversed(pila_expedientes)` recorre la lista **de atrás hacia adelante**, es decir, del expediente más reciente (la cima) al más antiguo. Esto es importante porque en una lista normal el último elemento agregado queda al final, pero conceptualmente representa el que está "arriba" de la pila.

```python
def consultar_pila():
    """Muestra todos los expedientes en espera (de arriba hacia abajo)."""
    if len(pila_expedientes) == 0:
        print("\nLa pila de expedientes esta vacia.\n")
        return

    print("\n=== Expedientes en espera (de arriba hacia abajo) ===")
    for expediente in reversed(pila_expedientes):
        numero = expediente["numero_expediente"]
        nombre = expediente["nombre"]
        edad = expediente["edad"]
        motivo = expediente["motivo_consulta"]
        print(f"  #{numero} | {nombre} ({edad} anios) - Motivo: {motivo}")
    print("=======================================================\n")
```

## Parte 5: Contar expedientes

La forma más simple de saber cuántos elementos tiene una pila (lista) es con la función incorporada `len()`. Esta función además **retorna** el total con `return total`, por si en el futuro quieres usar ese número en otra parte del programa (no solo imprimirlo).

```python
def contar_expedientes():
    """Cuenta cuantos expedientes hay actualmente en la pila."""
    total = len(pila_expedientes)
    print("\nHay " + str(total) + " expediente(s) en la pila.\n")
    return total
```

## Parte 6: Eliminar el elemento de la cima (pop)

Esta es la operación clásica de "desapilar":

- `pila_expedientes.pop()` **retira y retorna** el último elemento de la lista (la cima de la pila), sin necesidad de indicar una posición.
- Si la pila está vacía, `pop()` daría un error, por eso primero verificamos `len(pila_expedientes) == 0`.
- Se imprime la información del expediente eliminado, para que quede claro cuál se quitó.

```python
def eliminar_elemento_cima():
    """Elimina (pop) el elemento que esta en la cima de la pila."""
    if len(pila_expedientes) == 0:
        print("\nNo hay expedientes en la pila para eliminar.\n")
        return None

    expediente = pila_expedientes.pop()
    print("\n--- Expediente eliminado de la cima ---")
    print(f"Numero:  {expediente['numero_expediente']}")
    print(f"Nombre:  {expediente['nombre']}")
    print(f"Edad:    {expediente['edad']}")
    print(f"Motivo:  {expediente['motivo_consulta']}")
    print("----------------------------------------\n")
    return expediente
```

## Parte 7: Mostrar el menú

Esta función solo imprime en pantalla las opciones disponibles. Separarla del resto del programa hace que el código sea más ordenado: si algún día quieres cambiar el texto del menú, solo tocas esta función.

```python
def mostrar_menu():
    print("=========================================")
    print(" SISTEMA DE GESTION DE EXPEDIENTES (PILA)")
    print("=========================================")
    print("1. Anadir expediente")
    print("2. Eliminar toda la pila")
    print("3. Consultar expedientes")
    print("4. Contar expedientes")
    print("5. Eliminar elemento de la cima")
    print("6. Salir")
```

## Parte 8: Función principal (`main`)

Aquí se conecta todo lo anterior:

- `while True:` crea un bucle que se repite indefinidamente, mostrando el menú una y otra vez.
- `input(...)` captura la opción que escribe el usuario.
- La cadena de `if / elif` decide qué función llamar según la opción elegida.
- Cuando la opción es `"6"`, se imprime un mensaje de despedida y `break` **rompe el bucle**, terminando el programa.
- Si el usuario escribe algo que no es una opción válida (1-6), se le avisa y el menú vuelve a aparecer.

```python
def main():
    while True:
        mostrar_menu()
        opcion = input("Elige una opcion (1-6): ")

        if opcion == "1":
            anadir_expediente()
        elif opcion == "2":
            eliminar_toda_la_pila()
        elif opcion == "3":
            consultar_pila()
        elif opcion == "4":
            contar_expedientes()
        elif opcion == "5":
            eliminar_elemento_cima()
        elif opcion == "6":
            print("\nCerrando el programa. Hasta luego!")
            break
        else:
            print("\nOpcion no valida, intenta de nuevo.\n")
```

```python
main()
```
