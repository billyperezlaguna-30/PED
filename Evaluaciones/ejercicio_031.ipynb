## Instrucciones
Implementa la clase `CuentaBancariaV4` respetando los siguientes requerimientos:

1. **`__init__(self, nombre: str, saldo: float)`**:
   - Inicializa las variables de instancia: `nombre` (str), `saldo` (float), `transacciones` (list vacía) y `activo` (bool = True).

2. **`evaluar_condicion(self, limite: float, factor: float) -> float`**:
   - Usa operadores aritméticos, lógicos y condicionales (`if / elif / else`).
   - Si `saldo` >= limite y `activo` es True: retorna `saldo` * factor.
   - Elif `saldo` < limite y `saldo` > 0: retorna `saldo` + 10.0.
   - Else: retorna 0.0.

3. **`procesar_registros(self, datos: list) -> float`**:
   - Usa un ciclo `for` para recorrer `datos` (lista de floats/ints/None).
   - Ignora los elementos que sean `None`.
   - Acumula los valores válidos en `saldo` y agrega cada valor a `transacciones`.
   - Si `saldo` supera los 500.0, desactiva la entidad (`activo = False`).
   - Retorna el valor final de `saldo`.

4. **`acumular_hasta_objetivo(self, objetivo: float, paso: float) -> int`**:
   - Usa un ciclo `while` para incrementar `saldo` en `paso` mientras `saldo` sea estrictamente menor que `objetivo`.
   - Retorna la cantidad de iteraciones/pasos realizados.


```python
# Escribe tu solución aquí



class CuentaBancariaV4:
    def __init__(self, nombre: str, saldo: float):
        self.nombre = nombre
        self.saldo = saldo
        self.transacciones = []
        self.activo = True

    def evaluar_condicion(self, limite: float, factor: float) -> float:
        if self.saldo >= limite and self.activo:
            return self.saldo * factor
        elif self.saldo < limite and self.saldo > 0:
            return self.saldo + 10.0
        else:
            return 0.0
    def procesar_registros(self, datos: list) -> float:
        for registro in datos:
            if registro is not None:
                self.saldo += registro
                self.transacciones.append(registro)

        if self.saldo > 500.0:
            self.activo = False

        return self.saldo


    def acumular_hasta_objetivo(self, objetivo: float, paso: float) -> int:
        contador = 0
        while self.saldo < objetivo:
            self.saldo += paso
            contador += 1
        return contador
```
